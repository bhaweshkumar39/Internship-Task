## Clean Text Formatting




```python
text_columns = [
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "Country",
    "City",
    "Category",
    "Product",
    "Payment_Method"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()
```

##  Standardize Gender



```python
gender_map = {
    "male": "Male",
    "MALE": "Male",
    "M": "Male",
    "Female": "Female",
    "female": "Female",
    "FEMALE": "Female",
    "F": "Female"
}

df["Gender"] = df["Gender"].replace(gender_map)
df["Gender"] = df["Gender"].replace({
    "Malee": np.nan,
    "Fem": np.nan,
    "unknown": np.nan,
    "Unknown": np.nan
})

df["Gender"].value_counts(dropna=False)
```

## Standardize Country



```python
country_map = {
    "pakistan": "Pakistan",
    "PAKISTAN": "Pakistan",
    "India": "India",
    "india": "India",
    "INDIA": "India",
    "USA": "USA",
    "usa": "USA",
    "U.S.A": "USA",
    "United States": "USA",
    "UK": "UK",
    "uk": "UK",
    "U.K.": "UK",
    "United Kingdom": "UK",
    "Canada": "Canada",
    "canada": "Canada",
    "CANADA": "Canada",
    "UAE": "UAE",
    "uae": "UAE",
    "U.A.E.": "UAE",
    "United Arab Emirates": "UAE",
    "Pak": "Pakistan",
    "Pak ": "Pakistan",
    "USAa": "USA"
}

df["Country"] = df["Country"].replace(country_map)
df["Country"] = df["Country"].replace(
    ["Unknown", "unknown", "N/A", "NIL", "Invalid", "?"],
    np.nan
)

df["Country"].value_counts(dropna=False)
```

## Standardize Category and Payment Method




```python
df["Category"] = (
    df["Category"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({
        "electronics": "Electronics",
        "clothing": "Clothing",
        "home & kitchen": "Home & Kitchen",
        "books": "Books",
        "sports": "Sports",
        "unknown": np.nan,
        "n/a": np.nan,
        "other": np.nan,
        "x": np.nan,
        "invalid": np.nan,
        "-": np.nan,
        "none": np.nan,
        "?": np.nan
    })
)

payment_map = {
    "cash": "Cash",
    "credit card": "Credit Card",
    "debit card": "Debit Card",
    "paypal": "PayPal",
    "bank transfer": "Bank Transfer"
}

df["Payment_Method"] = (
    df["Payment_Method"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace(payment_map)
)

df["Payment_Method"] = df["Payment_Method"].replace(
    ["unknown", "n/a", "nil", "invalid", "?"],
    np.nan
)
```

##  Standardize Names and Cities



```python
df["Customer_Name"] = (
    df["Customer_Name"]
    .astype("string")
    .str.strip()
    .str.title()
)

df["City"] = (
    df["City"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

## Correct Numeric Data Types




```python
df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

df["Unit_Price"] = (
    df["Unit_Price"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["Unit_Price"] = pd.to_numeric(
    df["Unit_Price"], errors="coerce"
)

df["Rating"] = pd.to_numeric(
    df["Rating"], errors="coerce"
)

df["Discount"] = pd.to_numeric(
    df["Discount"], errors="coerce"
)

df["Total_Amount"] = pd.to_numeric(
    df["Total_Amount"], errors="coerce"
)
```

```python
df.loc[df["Quantity"] <= 0, "Quantity"] = np.nan
df.loc[df["Unit_Price"] <= 0, "Unit_Price"] = np.nan
df.loc[~df["Rating"].between(1, 5), "Rating"] = np.nan
df.loc[~df["Discount"].between(0, 1), "Discount"] = np.nan
df.loc[df["Total_Amount"] < 0, "Total_Amount"] = np.nan
```

## 14. Correct Date Format

The dataset contains several date formats. We convert the date column into one consistent datetime format.

`dayfirst=True` helps interpret dates such as `25/03/2024`.


```python
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce",
    dayfirst=True
)

df["Order_Date"].head()
```

## Remove Duplicate Rows




```python
duplicates_removed = df.duplicated().sum()

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", duplicates_removed)
print("Rows after duplicate removal:", len(df))
```

```python
categorical_columns = [
    "Customer_Name",
    "Gender",
    "Country",
    "City",
    "Category",
    "Product",
    "Payment_Method"
]

numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Rating",
    "Discount",
    "Total_Amount"
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

df["Order_Date"] = df["Order_Date"].ffill()

print("Remaining missing values:")
print(df.isnull().sum())
```

## Correct Final Data Types




```python
df["Customer_ID"] = (
    df["Customer_ID"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df["Quantity"] = df["Quantity"].round().astype("int64")
df["Unit_Price"] = df["Unit_Price"].astype("float64")
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Rating"] = df["Rating"].round().astype("int64")
df["Discount"] = df["Discount"].astype("float64")
df["Total_Amount"] = df["Total_Amount"].astype("float64")
```

## Validate the Cleaned Dataset



```python
print("Shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
print("Quantity range:", df["Quantity"].min(), "to", df["Quantity"].max())
print("Unit Price range:", df["Unit_Price"].min(), "to", df["Unit_Price"].max())
print("Rating range:", df["Rating"].min(), "to", df["Rating"].max())
print("Discount range:", df["Discount"].min(), "to", df["Discount"].max())
print("Total Amount range:", df["Total_Amount"].min(), "to", df["Total_Amount"].max())
```
